🧠 Conceptual Flow
OpenCV → Heavy-duty image processing (filters, edge detection, face recognition, etc.)--->Use BGR format
Pillow → Lightweight image manipulation (open, save, rotate, convert modes).
Matplotlib → Visualization (show images, plots, histograms).--->Use RGB format
NumPy → Backbone for pixel-level math (images = arrays).

```python
# 8)Write a Python statement to flip the image "car.jpg" left-to-right using OpenCV and display the result.

img = cv2.imread("car.jpeg")
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title(" without Flipped")
plt.axis("off")
plt.show()
flipped = cv2.flip(img, 1)  # Flip the image horizontally (left-to-right)

plt.imshow(cv2.cvtColor(flipped, cv2.COLOR_BGR2RGB))
plt.title("Flipped Left-to-Right")
plt.axis("off")
plt.show()
```

```python
 # 9) Write a Python program that converts an image from RGB to HSV colour space, and display each channel (Hue, Saturation, Value) separately using Matplotlib.

import cv2
import matplotlib.pyplot as plt

# Step 1: Load the image
img = cv2.imread("flower.jpg")   # Replace with your filename

# Step 2: Convert from BGR (OpenCV default) to HSV
hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)

# Step 3: Split HSV into individual channels
h, s, v = cv2.split(hsv)

# Step 4: Display each channel separately
plt.figure(figsize=(12,4))

plt.subplot(1,3,1)
plt.imshow(h, cmap='gray')
plt.title("Hue Channel")
plt.axis("off")

plt.subplot(1,3,2)
plt.imshow(s, cmap='gray')
plt.title("Saturation Channel")
plt.axis("off")

plt.subplot(1,3,3)
plt.imshow(v, cmap='gray')
plt.title("Value Channel")
plt.axis("off")

plt.show()
```
