```python
import cv2
import numpy as np
from matplotlib import pyplot as plt
from PIL import Image
```

```python
pil_img = Image.open("car.jpeg")
plt.imshow(pil_img)
plt.title("BBT")
plt.axis('off')
plt.show()
```

```python
cv_img = cv2.imread("car.jpeg")
cv_img_rgb = cv2.cvtColor(cv_img, cv2.COLOR_BGR2RGB)
plt.imshow(cv_img_rgb)
plt.title("Converted image COLOR_BGR2RGB")
plt.axis('off')
plt.show()
```

```python
gray = cv2.cvtColor(cv_img, cv2.COLOR_BGR2GRAY)
plt.imshow(gray, cmap='gray')
plt.title("Grayscale Image")
plt.axis('off')
plt.show()
```

```python
# Write a Python program to compute and display the histogram of a grayscale image using NumPy (manual method), without using cv2.calcHist().
# Manual Histogram (NumPy)
hist_manual, bins = np.histogram(gray.flatten(), 256, [0, 256])
plt.plot(hist_manual, color='black')
plt.title("Manual Histogram (NumPy)")
plt.xlabel("Intensity Values")
plt.ylabel("Frequency")
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

gray = cv2.imread("car.jpeg", cv2.IMREAD_GRAYSCALE)
hist_manual, bins = np.histogram(gray.flatten(), 256, [0, 256])

plt.plot(hist_manual, color='black')
plt.title("Manual Histogram (NumPy)")
plt.xlabel("Intensity")
plt.ylabel("Frequency")
plt.show()
```

```python
r_min, r_max = np.min(gray), np.max(gray)
stretched = ((gray - r_min) * (255 / (r_max - r_min))).astype(np.uint8)

plt.figure(figsize=(12,8))
plt.subplot(1,2,1)
plt.imshow(gray, cmap='gray')
plt.title("Original")
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(stretched, cmap='gray')
plt.title("Contrast Stretched")
plt.axis('off')
plt.show()
```
